```python
import numpy as np
```

```python
x=[1,0,3,2]
h=[1,1,1]
def convolve_direct(x, h):
    N = len(x)
    M = len(h)
    L = N+M-1
    y = np.zeros(L)

    for n in range(L):
        soma = 0
        for k in range(N):
            idx_h = n-k
            if 0 <= idx_h < M:
                soma = soma + x[k] * h[idx_h]
        y[n] = soma

    return y

y = convolve_direct(x, h)
print(y)

y_numpy = np.convolve(x, h)
print(y_numpy)
print(np.allclose(y, y_numpy))

x2 = np.array([2, 4, -1, 5, 3, 0, 7])   # N=7
h2 = np.array([1, -1])                   # M=2

y2 = convolve_direct(x2, h2)
y2_numpy = np.convolve(x2, h2)

print(y2)
print(y2_numpy)
print(np.allclose(y2, y2_numpy))
```

```python
import time

for N in [500, 1000, 2000]:
    x_teste = np.random.randn(N)
    h_teste = np.random.randn(N)

    inicio = time.perf_counter()
    y_teste = convolve_direct(x_teste, h_teste)
    fim = time.perf_counter()

    print(f"N={N} -> tempo: {fim - inicio:.4f} segundos")
```
