```python
import numpy as np

def dft(x):
    x = np.asarray(x, dtype=complex)
    N = len(x)
    X = np.zeros(N, dtype=complex)

    for k in range(N):
        soma = 0
        for n in range(N):
            soma += x[n] * np.exp(-1j * 2 * np.pi * k * n / N)
        X[k] = soma

    return X


# teste
x = [1, 1, 1, 1]
print(dft(x))
print(np.fft.fft(x))  # resultado de referência do numpy

print(np.allclose(dft(x), np.fft.fft(x)))
```

```python
def idft(X):
    X = np.asarray(X, dtype=complex)
    N = len(X)
    x = np.zeros(N, dtype=complex)

    for n in range(N):
        soma = 0
        for k in range(N):
            soma += X[k] * np.exp(1j * 2 * np.pi * k * n / N)
        x[n] = soma / N

    return x


# teste de ida e volta
x = [3, 5, 6, 7]
X = dft(x)
x_rec = idft(X)

print(np.round(x_rec, 4))
print(np.allclose(x_rec, x))
print(np.allclose(idft(X), np.fft.ifft(X)))  # compara com o numpy
```
