```python
import numpy as np

def dft(x):
    x = np.asarray(x, dtype=complex)
    N = len(x)
    X = np.zeros(N, dtype=complex)

    for k in range(N):
        soma = 0
        for n in range(N):
            soma += x[n] * np.exp(-1j * 2 * np.pi * k * n / N)
        X[k] = soma

    return X


# teste
x = [1, 1, 1, 1]
print(dft(x))
print(np.fft.fft(x))  # resultado de referência do numpy

print(np.allclose(dft(x), np.fft.fft(x)))
```

```python
def idft(X):
    X = np.asarray(X, dtype=complex)
    N = len(X)
    x = np.zeros(N, dtype=complex)

    for n in range(N):
        soma = 0
        for k in range(N):
            soma += X[k] * np.exp(1j * 2 * np.pi * k * n / N)
        x[n] = soma / N

    return x


# teste de ida e volta
x = [3, 5, 6, 7]
X = dft(x)
x_rec = idft(X)

print(np.round(x_rec, 4))
print(np.allclose(x_rec, x))
print(np.allclose(idft(X), np.fft.ifft(X)))  # compara com o numpy
```

```python
import numpy as np

rng = np.random.default_rng(0)
N = 8
x = rng.standard_normal(N)
y = rng.standard_normal(N)
X = dft(x)
Y = dft(y)
n = np.arange(N)
k = np.arange(N)

# 1. linearidade: DFT(a*x + b*y) = a*X + b*Y
a, b = 2, -3
print("Linearidade:", np.allclose(dft(a*x + b*y), a*X + b*Y))

# 2. deslocamento circular no tempo: x[(n-m) mod N] -> X[k] * e^(-j2πkm/N)
m = 3
print("Deslocamento no tempo:", np.allclose(dft(np.roll(x, m)), X * np.exp(-1j*2*np.pi*k*m/N)))

# 3. deslocamento em frequência: x[n] * e^(j2πln/N) -> X[(k-l) mod N]
l = 2
print("Deslocamento em frequência:", np.allclose(dft(x * np.exp(1j*2*np.pi*l*n/N)), np.roll(X, l)))

# 4. reversão circular no tempo: x[(-n) mod N] -> X[(-k) mod N]
print("Reversão no tempo:", np.allclose(dft(x[(-n) % N]), X[(-k) % N]))

# 5. simetria conjugada (x real): X[(-k) mod N] = conj(X[k])
print("Simetria conjugada:", np.allclose(X[(-k) % N], np.conj(X)))

# 6. convolução circular: DFT(x ⊛ y) = X * Y
def conv_circular(x, y):
    N = len(x)
    z = np.zeros(N)
    for n in range(N):
        for m in range(N):
            z[n] += x[m] * y[(n - m) % N]
    return z

print("Convolução circular:", np.allclose(dft(conv_circular(x, y)), X * Y))

# 7. Parseval: energia no tempo = energia na frequência / N
print("Parseval:", np.allclose(np.sum(np.abs(x)**2), np.sum(np.abs(X)**2) / N))
```

```python
def conv_linear_dft(x, h):
    x = np.asarray(x, dtype=float)
    h = np.asarray(h, dtype=float)
    L = len(x) + len(h) - 1

    # zero-padding até o tamanho L
    xp = np.concatenate([x, np.zeros(L - len(x))])
    hp = np.concatenate([h, np.zeros(L - len(h))])

    # convolução = multiplicação na frequência
    y = idft(dft(xp) * dft(hp))

    # o resultado deve ser real; a parte imaginária é só erro de arredondamento
    return np.real(y)


x = [1, 2, 3, 4]
h = [1, 1, 1]

print("Com zero-padding:", np.round(conv_linear_dft(x, h), 4))
print("np.convolve:     ", np.convolve(x, h))
print(np.allclose(conv_linear_dft(x, h), np.convolve(x, h)))

# sem zero-padding: convolução circular, resultado errado (aliasing)
h_mesmo_tamanho = np.concatenate([h, np.zeros(len(x) - len(h))])
sem_padding = np.real(idft(dft(x) * dft(h_mesmo_tamanho)))
print("Sem zero-padding:", np.round(sem_padding, 4))
```
